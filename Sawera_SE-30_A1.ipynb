```python
students = []
def result(marks):
    total = sum(marks)
    percentage = total / 3
    if percentage >= 80:
        grade = "A"
    elif percentage >= 70:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 50:
        grade = "D"
    else:
        grade = "F"
    return total, percentage, grade
for i in range(5):
    name = input("enter student name:")
    roll = input("enter roll number:")
    python = int(input("enter python marks:"))
    ai = int(input("enter AI marks:"))
    math = int(input("enter math marks:"))
    marks = [python, ai, math]
    total, percentage, grade = result(marks)
    student = {
        "name": name,
        "roll": roll,
        "marks": marks,
        "total": total,
        "percentage": percentage,
        "grade": grade
    }
    students.append(student)
print("student results")

for student in students:
    print("name:", student["name"])
    print("roll no.:", student["roll"])
    print("total:", student["total"])
    print("percetage:", student["percentage"])
    print("grade:", student["grade"]) 
highest = students[0]

for student in students:
    if student["percentage"] > highest["percentage"]:
        highest = student
print("highest percentage")
print("name:", highest["name"])
print("percentage:", highest["percentage"])
file = open("results.txt", "w")
for student in students:
    file.write("name: " + student["name"] )
    file.write("roll no.: " + student["roll"] )
    file.write("total: " + str(student["total"]) )
    file.write("percentage: " + str(student["percentage"]) )
    file.write("graade: " + student["grade"] )
file.close()
```

```python
def analysis(sentence):
    words = sentence.split()
    print("total words:", len(words))
    print("yotal characters:", len(sentence))
    longest = words[0]
    shortest = words[0]
    for word in words:
        if len(word) > len(longest):
            longest = word
        if len(word) < len(shortest):
            shortest = word
    print("Longest word:", longest)
    print("Shortest word:", shortest)
    vowels = 0
    for ch in sentence:
        if ch.lower() in "aeiou":
            vowels += 1
    print("No of vowels:", vowels)
    unique = set(words)
    print("Unique words:", unique)
    print("alphabetical order:")
    alphabet = sorted(unique)
    for word in alphabet:
        print(word)

    file = open("word_analysis.txt", "w")
    file.write("total words: " + str(len(words)))
    file.write("yotal characters: " + str(len(sentence)))
    file.write("longest word: " + longest)
    file.write("shortest word: " + shortest)
    file.write("No of vowels: " + str(vowels))
    file.write("nique words: " + str(unique))
    file.write("alphabetical words: " + str(alphabet))
    file.close()

sentence = input("enter a sentence: ")

analysis(sentence)
```

```python
products = {}
def bill(products):
    total = 0
    for product in products:
        price = products[product]["price"]
        quantity = products[product]["quantity"]
        products[product]["total"] = price * quantity
        total = total + products[product]["total"]
    return total
for i in range(5):
    name = input("enter product name: ")
    price = int(input("enter price: "))
    quantity = int(input("enter quantity: "))
    products[name] = {
        "price": price,
        "quantity": quantity
    }

total = bill(products)
discount = 0
if total > 10000:
    discount = total * 0.10
final_bill = total - discount
expensive = ""
for product in products:
    if expensive == "":
        expensive = product
    elif products[product]["price"] > products[expensive]["price"]:
        expensive = product
print("shopping Bill")
for product in products:
    print("product:", product)
    print("price:", products[product]["price"])
    print("quantity:", products[product]["quantity"])
    print("total:", products[product]["total"])
print("complete bill:", total)
print("discount:", discount)
print("final bill:", final_bill)
print("most expensve product:", expensive)
file = open("bill.txt", "w")
for product in products:
    file.write("product: " + product)
    file.write("price: " + str(products[product]["price"]))
    file.write("quantity: " + str(products[product]["quantity"]))
    file.write("total: " + str(products[product]["total"]))
file.write("complete bill: " + str(total))
file.write("discount: " + str(discount))
file.write("final bill: " + str(final_bill))
file.write("most expensive product: " + expensive)
file.close()
```

```python
def addition(a, b):
    return a + b
def subtraction(a, b):
    return a - b
def multiplication(a, b):
    return a * b
def division(a, b):
    return a / b
def modulus(a, b):
    return a % b
while True:
    a = int(input("enter first number: "))
    b = int(input("enter second number: "))
    print("+ for addition")
    print("- for subtraction")
    print("* for multiplication")
    print("/ for division")
    print("% for modulus")
    operation=input("enter operation: ")
    if operation=="+":
        print("result:", addition(a, b))
    elif operation=="-":
        print("result:", subtraction(a, b))
    elif operation=="*":
        print("result:", multiplication(a, b))
    elif operation=="/":
        if b != 0:
            print("result:", division(a, b))
        else:
            print("cannot divide by zero")
    elif operation=="%":
        if b != 0:
            print("result:", modulus(a, b))
        else:
            print("cannot divide by zero")
    else:
        print("invalid operation")
    again = input("do you want another calculation? yes/no: ")
    if again == "no":
        break
```

```python
menu = {
    1: ["Burger", 500],
    2: ["Pizza", 800],
    3: ["Fries", 250],
    4: ["Drink", 150]
}
def calculate_bill(order):
    total = 0
    for item in order:
        price = menu[item][1]
        quantity = order[item]
        total = total + price * quantity
    return total
order = {}
while True:
    print("Menu")
    print("1.burger Rs. 500")
    print("2.pizza Rs. 800")
    print("3.fries Rs. 250")
    print("4.drink Rs. 150")
    print("5.exit")
    choice = int(input("enter your choice:"))
    if choice == 5:
        break
    if choice in menu:
        quantity = int(input("enter quantity:"))
        order[choice] = quantity
        print("item added")
    else:
        print("invalid choice")
total = calculate_bill(order)
print("final bill")
for item in order:
    name = menu[item][0]
    price = menu[item][1]
    quantity = order[item]
    print(name, quantity, price * quantity)
print("total bill:", total)
file = open("orders.txt", "w")
file.write("restaurant ill")
for item in order:

    name = menu[item][0]
    price = menu[item][1]
    quantity = order[item]
    file.write(name + " " + str(quantity) + " " + str(price * quantity))
file.write("total bill: " + str(total))
file.close()
```
